**Phần 4**

```python
text_p4_1 = "The cat chased the mouse and the dog watched them."
doc_p4_1 = nlp(text_p4_1)

for token in doc_p4_1:
    if token.pos_ == "VERB":
        verb = token.text
        subject = ""
        obj = ""
        for child in token.children:
            if child.dep_ == "nsubj":
                subject = child.text
            if child.dep_ == "dobj":
                obj = child.text
        if subject and obj:
            print(f"Found Triplet: ({subject}, {verb}, {obj})")
```

```python
text_p4_2 = "The big, fluffy white cat is sleeping on the warm mat."
doc_p4_2 = nlp(text_p4_2)

for token in doc_p4_2:
    if token.pos_ == "NOUN":
        adjectives = []
        for child in token.children:
            if child.dep_ == "amod":
                adjectives.append(child.text)
        if adjectives:
            print(f"Danh từ '{token.text}' được bổ nghĩa bởi các tính từ: {adjectives}")
```

**Phần 5**

```python
import spacy

def find_main_verb(doc):
    for token in doc:
        if token.dep_ == "ROOT":
            return token
    return None

nlp = spacy.load("en_core_web_sm")

text = "The cat chased the mouse."
doc = nlp(text)
main_verb = find_main_verb(doc)
print(f"Động từ chính: {main_verb.text}")
```

```python
def extract_noun_chunks(doc):
    noun_chunks = [chunk.text for chunk in doc.noun_chunks]
    return noun_chunks

text = "The big, fluffy white cat is sleeping on the warm mat."
doc = nlp(text)
chunks = extract_noun_chunks(doc)
print(chunks)
```

```python
def get_path_to_root(token):
    path = [token.text]
    current_token = token

    while current_token.dep_ != "ROOT":
        current_token = current_token.head
        path.append(current_token.text)
    return path

text = "The quick brown fox jumps over the lazy dog."
doc = nlp(text)
token_dog = doc[-2]
path = get_path_to_root(token_dog)
print(f"Đường đi từ '{token_dog.text}' lên ROOT: {' -> '.join(path)}")
```
